```python
import pandas as pd
```

```python
bprk = pd.read_csv('Full1024.csv')
bprs = pd.read_csv('Full1024BPRS.csv')
```

```python
map = pd.read_csv('BankLongLat.csv')
map
```

```python
# Check columns in bprk and bprs
print("BPRK columns:", bprk.columns.tolist())
print("\nBPRS columns:", bprs.columns.tolist())
print("\nMap columns:", map.columns.tolist())
```

```python
bprs
```

```python
# Drop kolom Latitude dan Longitude yang lama, ambil dari map
bprk = bprk.drop(columns=['Latitude', 'Longitude'], errors='ignore')
bprs = bprs.drop(columns=['Latitude', 'Longitude'], errors='ignore')

# Merge bprk dengan map berdasarkan nama bank
bprk = pd.merge(
    bprk, 
    map[['Nama Bank', 'Latitude', 'Longitude']], 
    left_on='Nama_BPR', 
    right_on='Nama Bank', 
    how='left'
)

# Merge bprs dengan map berdasarkan nama bank
bprs = pd.merge(
    bprs, 
    map[['Nama Bank', 'Latitude', 'Longitude']], 
    left_on='Nama_BPR', 
    right_on='Nama Bank', 
    how='left'
)

# Drop kolom 'Nama Bank' karena udah ada 'Nama_BPR'
bprk = bprk.drop(columns=['Nama Bank'], errors='ignore')
bprs = bprs.drop(columns=['Nama Bank'], errors='ignore')

print(f"BPRK: {len(bprk)} rows")
print(f"BPRS: {len(bprs)} rows")
```
